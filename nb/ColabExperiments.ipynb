```python
print(witness_v)
```

### Caricare i testimoni nell'istanza di collazione

```python
collation_1.add_plain_witness('P', witness_p)
collation_1.add_plain_witness('V', witness_v)
collation_1.add_plain_witness('L', witness_l)
```

### Allineamento

```python
alignment_table_ns = collate(collation_1, layout='vertical', segmentation=False)
print(alignment_table_ns)
```
